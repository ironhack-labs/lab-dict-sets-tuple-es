#### Ahora, usa un bucle FOR para verificar si cada letra en la siguiente lista está presente en `tup3`:


```
letters = ["a", "b", "c", "d", "e"]
```


Para cada letra que verifiques, imprime `True` si está presente en `tup3`, de lo contrario imprime `False`.

*Sugerencia: solo necesitas hacer un bucle con `letters`. No necesitas hacer un bucle con `tup3` porque hay un operador de Python `in` que puedes usar. Consulta la [referencia](https://stackoverflow.com/questions/17920147/how-to-check-if-a-tuple-contains-an-element-in-python).*

````python
# Tu codígo aquí
letters = ["a", "b", "c", "d", "e"]

for characters in letters:
    if characters in tup3:
        print("True")
    else:
        print("False")
        
    
````

#### ¿Cuántas veces aparece cada letra de `letters` en `tup3`?

Imprime el número de ocurrencias de cada letra.

````python
# Tu codígo aquí
# Creamos un diccionario para almacenar los resultados
char_count = {}
# Inicializamos el contador para cada letra en letters
for letter in letters:
    char_count[letter] = 0
# Iteración sobre cada elemento de la tupla
for element in tup3:
    #Si está en letters, añadimos al contador
    if element in letters:
        char_count[element] += 1
#imprimimos resultados
print(char_count)
````

#### En la celda de abajo, crea una lista llamada `sample_list_1` con 80 valores aleatorios.

Requisitos:

* Cada valor es un entero entre 0 y 100.
* Cada valor en la lista es único.

Imprime `sample_list_1` para revisar sus valores.

*Sugerencia: usa `random.sample` ([referencia](https://docs.python.org/3/library/random.html#random.sample)).*

````python
# tu codigo aquí
import random

num_unicos = set()

list_size = 80
min_range = 0
max_range = 100

while len(num_unicos) < list_size:
    number = random.randint(min_range, max_range) 
    num_unicos.add(number)

sample_list_1 = list(num_unicos)
print(sample_list_1)
````

#### Convierte `sample_list_1` en un conjunto llamado `set1`. Imprime la longitud del conjunto. ¿Su longitud sigue siendo 80?

````python
# tu codigo aquí
set1 = set(sample_list_1)
print(set1)

long = len(set1)
print("La longitud del set sigue siendo de", long)
````

#### Crea otra lista llamada `sample_list_2` con 80 valores aleatorios.

Requisitos:

* Cada valor es un entero entre 0 y 100.
* Los valores en la lista no tienen que ser únicos.

*Sugerencia: Usa un bucle FOR.*

````python
# tu codigo aquí
import random
sample_list_2 = []

for _ in range(80):
    sample_list_2.append(random.randint(0,100))

print(sample_list_2)
````

#### Convierte `sample_list_2` en un conjunto llamado `set2`. Imprime la longitud del conjunto. ¿Su longitud sigue siendo 80?

````python
# tu codigo aquí
set2 = set(sample_list_2)
print(set2)

longg = len(set2)
print("La longitud del set es de", longg)
#La longitud del set ya no es de 80 porque se han eliminado números duplicados.
````

#### Ordena los valores de `word_freq` de forma ascendente.

Ordenar los valores de un diccionario es más complicado que ordenar las claves porque los valores de un diccionario no son únicos. Por lo tanto, no puedes usar la misma forma en que ordenaste las claves del diccionario para ordenar los valores del diccionario.

La forma de ordenar un diccionario por valor es utilizar las funciones `sorted` y `operator.itemgetter`. El siguiente fragmento de código se te proporciona para que lo pruebes. Te dará una lista de tuplas en la que cada tupla contiene la clave y el valor de un elemento del diccionario. Y la lista está ordenada basada en el valor del diccionario ( [referencia](http://thomas-cokelaer.info/blog/2017/12/how-to-sort-a-dictionary-by-values-in-python/)
).

```python
import operator
sorted_tups = sorted(word_freq.items(), key=operator.itemgetter(1))
print(sorted_tups)
```

Por lo tanto, los pasos para ordenar `word_freq` por valor son:

* Itera la lista de tuplas. Inserta cada par clave-valor en `word_freq2` como un objeto.
